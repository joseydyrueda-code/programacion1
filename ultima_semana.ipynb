escritura de archivos desde python

python permite cargar y crear objetos de archivos dados, y tambien permite crear y editar archivos dados

```r
#crear y escribir un archivo de texto
with open("archivo.txt", "w") as f:
  f.write("hola, este es un archivo de texto.\n")
  f.write("segunda linea.\n")
```

con el argumento "w" se crea el archivo si no existe. tambien sobreescribe el archivo si ya existe, borrando su contenido previo

con archivos .txt (se tiene que tener en cuenta la forma de separar espacios)



```r
with open("archivo.txt", "r") as f: #vamos a leer el objeto con la funcion r (ademas de llamar al objeto como f)
  print(f.read())
```

con el argumento "r" se leen archivos
