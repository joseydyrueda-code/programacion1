```python
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}
try:
  usuario = input("usuarios: ").upper()
  password = input("contraseñas: ")

  if usuario not in usuarios:
    raise KeyError("El ususario no existe")

  if usuarios[usuario] != password:
      raise ValueError ("contraseña incorrecta")

  else:
        print("bienvenido,", usuario)
except KeyError as e:
   print("Error:", e)

except ValueError as e:
   print("Error:", e)
```

```python
try:
    num1 = float(input("ingrese el primer numero: "))
    num2 = float(input("ingrese el segundo numero: "))
    op = input("ingrese la operacion (+, *, -, /, %): ")

    if op == "+":
        print("la suma es:", num1 + num2)
    elif op == "-":
        print("la resta es:", num1 - num2)
    elif op == "*":
        print("la multiplicacion es:", num1 * num2)
    elif op == "/":
        # Manejo simple para evitar que colapse por ZeroDivisionError
        if num2 == 0:
            raise ZeroDivisionError("no se puede dividir por cero")
        print("la division es:", num1 / num2)
    elif op == "%":
        print("el residuo es:", num1 % num2)
    else:
        raise ValueError("operacion no valida")

except (ValueError, ZeroDivisionError) as e:
    print(f"Error detectado: {e}")
    print("Cerrando el programa...")
```

```python
productos = {
    "pan": 1.5,
    "leche": 2.0,
    "huevos": 3.0
}

print("Productos disponibles:", productos)

try:
  nombre = input("¿Qué producto quieres?: ").lower()
  if nombre not in productos:
    raise KeyError("El producto no existe")

  cantidad = int(input("Cantidad: "))
  if cantidad <= 0:
    raise ValueError("La cantidad debe ser mayor a 0")

  total = productos[nombre] * cantidad
  print("Total a pagar:", total)

except KeyError as e:
  print(f"Error:", e)
except ValueError as e:
  print(f"Error:", e)
```
