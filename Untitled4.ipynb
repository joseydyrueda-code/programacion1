```python
def sumar_matrices(A, B):
    filas = len(A)
    columnas = len(A[0])

    if filas != len(B) or columnas != len(B[0]):
        return "Error: Las matrices deben tener las mismas dimensiones para sumarse."

    resultado = []
    for i in range(filas):
        fila = []
        for j in range(columnas):
            fila.append(A[i][j] + B[i][j])
        resultado.append(fila)

    return resultado
```

```python
def contador_vocales1(palabra):
 vocales = "aeiou"
 consonantes = "bcdfghjklmnñpqrstvwxyz"
 contador = 0
 contador1 = 0
 contador2 = 0

 for letra in palabra.lower():
    if letra in vocales:
      contador += 1
    elif letra in consonantes:
         contador1 += 1
    else:
     contador2 += 1
 return({"Cantidad de vocales:": contador,
 "Cantidad de consonates:": contador1,
 "Cantidad de otros caracteres:": contador2})
```

funciones lambda:
son funciines anonimas y dee una sola expresion

```python
def sumar(a, b):
  return a + b
```

```python
suma = lambda a, b: a+b
suma(3, 4)
```

```python
sumar = lambda a, b: a + b
print(sumar(3, 5))
```

```python
multiplicar = lambda a, b: a*b
print(sumar(3, 5))
```

```python
ultimoelemento = lambda palabra: palabra[-1]
ultimoelemento("casa")
```

```python
def cuadrado(x):
  return x**2
```

```python
list(map(cuadrado, [1, 2, 3, 4]))
```

```python
list(map(lambda x: x**2, [1, 2, 3, 4]))
```

```python
def es_par(numero):
  return numero % 2 == 0
```

```python
es_par(5)
```

```python
es_par = lambda numero: numero % 2 == 0
es_par(5)
```

```python
list(filter(lambda numero: numero % 2 == 0, [1, 2, 3, 4]))
```
