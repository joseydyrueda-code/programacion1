```python
def sumar_matrices(A, B):
    filas = len(A)
    columnas = len(A[0])

    if filas != len(B) or columnas != len(B[0]):
        return "Error: Las matrices deben tener las mismas dimensiones para sumarse."

    resultado = []
    for i in range(filas):
        fila = []
        for j in range(columnas):
            fila.append(A[i][j] + B[i][j])
        resultado.append(fila)

    return resultado
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Limpiar datos
cerveza = tabla_alcohol1['cerveza'].dropna()

# Contar valores
conteo = cerveza.value_counts().sort_index()

# Estilo bonito
sns.set(style="whitegrid")

plt.figure(figsize=(12,7))

# Crear barras
bars = plt.bar(
    conteo.index.astype(str),
    conteo.values,
    color=sns.color_palette("coolwarm", len(conteo)),
    edgecolor="black"
)

# Agregar números encima de cada barra ✨
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.5,
        f'{int(height)} 🍺',
        ha='center',
        fontsize=11
    )

# Títulos con estilo 💖
plt.title('🍻 Consumo de Cerveza 🍻', fontsize=18, weight='bold')
plt.xlabel('Nivel de Consumo', fontsize=13)
plt.ylabel('Cantidad de Personas', fontsize=13)

# Rotar etiquetas si hace falta
plt.xticks(rotation=30)

# Quitar bordes innecesarios
sns.despine()

plt.tight_layout()
plt.show()
```
