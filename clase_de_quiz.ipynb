```python
class Persona:
  #construcctor
  def __init__(self, nombre, edad, estatura, peso):
    #aqui definimos los atibutos del obejeto
    #el self lo utilizamos para especificar que cada objeto tiene atributos diferentes
    #self significa "para el mismo objeto, en este caso, llamado persona"
    self.nombre = nombre #atributo
    self.edad = edad #atributo
    self.estatura = estatura #atributo
    self.peso = peso #atributo
#self guarda los atributos

  def imc(self):
    return self.peso / self.estatura ** 2

  def saludar(self):
    print("hola", self.nombre, "tu indice de masa corporal es: ", self.imc())
```

```python
class Calculadora:
  def __init__(self, a, b):
    self.a = a
    self.b = b
  def suma(self):
    return self.a + self.b
  def resta(self):
    return self.a - self.b
  def multiplicacion(self):
    return self.a * self.b
  def division(self):
    try:
      return self.a / self.b
    except ZeroDivisionError:
      print("no se puede dividir entre cero")
      return None
```

```python
calcula1 = Calculadora(2, 3)
calcula1.division()
```

```python
calcula2 = Calculadora(2, 0)
calcula2.division()
```

```python
calc = Calculadora(3,4)
print("suma", calc.suma())
print("resta", calc.resta())
print("multiplicacion", calc.multiplicacion())
print("division", calc.division())
```

```python
class Estudiante:
  """ metodos: aprobar"""
  def __init__(self, nombre, nota):
    self.nombre = nombre
    self.nota = nota

  def aprobar(self):
    if self.nota >= 3.0:
        print("el estudiante", self.nombre, "aprobó con nota", self.nota)
    elif self.nota <=2.0
        print("el estudiante", self.nombre, "es un bruto y reprobó con nota", self.nota)
    else:
        print("el estudiante", self.nombre, "reprobó con nota", self.nota)
```

```python
class CuentaBancaria:
  def __init__(self, titular, saldo):
    self.titular = titular
    self.__saldo = saldo #"__" le permite ser un atributo privado
    #metodo para mostrar saldo
    def mostrar_saldo(self):
      print("saldo:", self.__saldo)
      #metodo para depositar dinero
      def depositar(self, cantidad):
        if cantidad >0:
          self.__saldo += cantidad
```
